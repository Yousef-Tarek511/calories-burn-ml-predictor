# Decision Tree Regression

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## Importing the dataset

```python
dataset = pd.read_csv('/cleaned_dataset.csv')
dataset.head()
```

### Information about dataset

```python
dataset.shape
```

```python
dataset.info()
```

```python
dataset.describe().T
```

```python
dataset.describe(include=['object', 'string']).T
```

### Check Missing Value

```python
dataset.isnull().sum()
```

### Check Duplicates

```python
dataset.duplicated().sum()
```

```python
dataset.drop_duplicates(inplace=True)
```

### Split the dataset into Features and target

```python
X = dataset.iloc[:, :-1].values
y = dataset.iloc[:, -1].values
```

```python
print(X)
```

```python
print(y)
```

## Encoding categorical data

```python
# use get_dummies instead of above code
X = pd.get_dummies(dataset.iloc[:, :-1], drop_first=True , dtype=int).values
print(X)
```

## Splitting the dataset into the Training set and Test set

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 0)
```

```python
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

## Training the  Decision Tree Regression model on the Training set

```python
from sklearn.impute import SimpleImputer
from sklearn.tree import DecisionTreeRegressor
from sklearn.pipeline import make_pipeline

regressor = make_pipeline(
    SimpleImputer(strategy='median'),
    DecisionTreeRegressor(random_state=0)
)

regressor.fit(X_train, y_train)
```

## Predicting the Test set results

```python
y_pred = regressor.predict(X_test)
np.set_printoptions(precision=2)
print(np.concatenate((y_pred.reshape(len(y_pred),1), y_test.reshape(len(y_test),1)),1))
```

### Evaluate using MSE

```python
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error

mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mse)

print("Train R2 Score:", r2_score(y_train, regressor.predict(X_train)))
print("Test R2 Score:", r2)
print("Mean Squared Error:", mse)
print("Mean Absolute Error:", mae)
print("Root Mean Squared Error:", rmse)
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.pipeline import Pipeline

pipe = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('dt', DecisionTreeRegressor(random_state=0))
])

param_grid = {
    'dt__max_depth': [3, 5, 7, 10, 15, None],
    'dt__min_samples_split': [2, 5, 10, 20],
    'dt__min_samples_leaf': [1, 2, 5, 10]
}

grid_search = GridSearchCV(pipe, param_grid, cv=5, scoring='r2', n_jobs=-1)
grid_search.fit(X_train, y_train)

print("Best Parameters:", grid_search.best_params_)
print("Best CV R2 Score:", grid_search.best_score_)
```

```python
best_regressor = grid_search.best_estimator_
y_pred_best = best_regressor.predict(X_test)

mse_best = mean_squared_error(y_test, y_pred_best)
r2_best = r2_score(y_test, y_pred_best)
mae_best = mean_absolute_error(y_test, y_pred_best)
rmse_best = np.sqrt(mse_best)

print("Train R2 Score:", r2_score(y_train, best_regressor.predict(X_train)))
print("Test R2 Score:", r2_best)
print("Mean Squared Error:", mse_best)
print("Mean Absolute Error:", mae_best)
print("Root Mean Squared Error:", rmse_best)
```

```python
best_params = {k.replace('dt__', ''): v for k, v in grid_search.best_params_.items()}
test_sizes = [0.1, 0.2, 0.3, 0.4]

for size in test_sizes:
    X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=size, random_state=0)

    reg = make_pipeline(
        SimpleImputer(strategy='median'),
        DecisionTreeRegressor(random_state=0, **best_params)
    )
    reg.fit(X_tr, y_tr)
    pred = reg.predict(X_te)

    r2 = r2_score(y_te, pred)
    mae = mean_absolute_error(y_te, pred)
    mse = mean_squared_error(y_te, pred)
    rmse = np.sqrt(mse)

    print(f"--- Results for Test Size: {size * 100}% ---")
    print("R2 Score:", r2)
    print("MAE:", mae)
    print("MSE:", mse)
    print("RMSE:", rmse)
    print("-" * 40, "\n")
```
